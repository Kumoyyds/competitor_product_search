# This is to test the api of brightdata

```python
import dotenv
import os
import requests

dotenv.load_dotenv()
bright_data_key = os.getenv("BRIGHT_DATA_KEY")
```

```python
data_root = "./data/"
```

```python
def web_unlocker_html_extractor(target_url):
    api_url = "https://api.brightdata.com/request"

    payload = {
        "zone": "web_unlocker1",
        "url": target_url,
        "format": "raw",
        "method": "GET",
    #  "country": "de",
    }
    headers = {
        "Authorization": f"Bearer {bright_data_key}",
        "Content-Type": "application/json",
    }

    response = requests.post(api_url, json=payload, headers=headers)
    print(response.status_code)
    return {'response_status': response.status_code, 'response_text': response.text}
    
```
